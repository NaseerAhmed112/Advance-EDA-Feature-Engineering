## Step 5.1: Create a Backup of the Original Dataset

```python
numeric_columns = df.select_dtypes(include=['int64', 'float64']).columns

categorical_columns = df.select_dtypes(
    include=['object', 'string', 'category']
).columns

print("Numeric Columns:")
print(numeric_columns)

print("\nCategorical Columns:")
print(categorical_columns)
```

```python
df_mean = df.copy()

for col in numeric_columns:

    df_mean[col] = df_mean[col].fillna(df_mean[col].mean())

print("Mean Imputation Completed!")
```

```python
df_mean[numeric_columns].isnull().sum()
```

```python
## Step 5.4: Median Imputation (Recommended)
```

```python
df_median = df.copy()

for col in numeric_columns:

    df_median[col] = df_median[col].fillna(df_median[col].median())

print("Median Imputation Completed!")
```

```python

# Step 5.5: Mode Imputation


df_mode = df_median.copy()

for col in categorical_columns:

    df_mode[col] = df_mode[col].fillna(df_mode[col].mode()[0])

print("Mode Imputation Completed!")
```

# Step 6.1: Detect Outliers using IQR Method

```python

# Step 6.1: Detect Outliers using IQR Method

numeric_cols = ['Quantity', 'UnitPrice', 'ItemsInCart', 'TotalPrice']

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"{col} --> {len(outliers)} Outliers")
```

## Step 6.2: Visualize Outliers

```python
for col in numeric_cols:
    plt.figure(figsize=(6,4))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.show()
```

## Step 6.3: Remove Outliers

```python
df_clean = df.copy()

for col in numeric_cols:

    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df_clean = df_clean[
        (df_clean[col] >= lower) &
        (df_clean[col] <= upper)
    ]

print("Dataset Shape Before:", df.shape)
print("Dataset Shape After :", df_clean.shape)
```
